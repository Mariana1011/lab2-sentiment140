```python
import requests
import json

API_BASE = 'http://34.204.240.213:8000'

print('API:', API_BASE)
```

## Health

```python
health = requests.get(f'{API_BASE}/health').json()
health
```

## Protocolo experimental

```python
protocol = requests.get(f'{API_BASE}/audit/protocol').json()
protocol
```

## Modelo champion

```python
model = requests.get(f'{API_BASE}/audit/model').json()
model
```

## Contribuciones

```python
contributions = requests.get(
    f'{API_BASE}/audit/contributions'
).json()

contributions
```

```python
for member in contributions['members']:
    print(
        member['member_id'],
        'configuraciones:', member['valid_configurations'],
        'etapas:', member['stage_count'],
        'cumple:', member['meets_requirement']
    )

print('invalid_run_ids:', contributions['invalid_run_ids'])
print('unattributed_run_ids:', contributions['unattributed_run_ids'])
```

## Runs registrados

```python
runs = requests.get(f'{API_BASE}/audit/runs').json()

print('Número de runs presentados:', len(runs['runs']))
```
